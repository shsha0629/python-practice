```python
import keras
import tensorflow as tf
from sklearn.model_selection import train_test_split
```

```python
(train_input, train_target), _ = keras.datasets.fashion_mnist.load_data()

train_scaled = train_input / 255.0
train_scaled, val_scaled, train_target, val_target = train_test_split(
    train_scaled, train_target, test_size=0.2, random_state=42
)

def model_fn(a_layer=None):
    model = keras.Sequential()
    model.add(keras.layers.Input(shape=(28, 28)))
    model.add(keras.layers.Flatten())
    model.add(keras.layers.Dense(100, activation='relu'))
    if a_layer:
        model.add(a_layer)
    model.add(keras.layers.Dense(10, activation='softmax'))
    
    return model
```

```python
for dr in [0.0, 0.2, 0.3, 0.5]:
    keras.utils.set_random_seed(42)

    model = model_fn(a_layer=keras.layers.Dropout(dr))

    model.compile(optimizer='adam',
                loss=keras.losses.sparse_categorical_crossentropy,
                metrics=[keras.metrics.sparse_categorical_accuracy])
    history = model.fit(train_scaled, train_target,
                        epochs=20, verbose=0,
                        validation_data=(val_scaled, val_target))
    
    train_acc = history.history['sparse_categorical_accuracy'][-1]
    val_acc = history.history['val_sparse_categorical_accuracy'][-1]
    print(f"Dropout = {dr} : train_acc = {train_acc:.4f} | val_acc = {val_acc:.4f}")
```

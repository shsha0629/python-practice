```python
import keras
import tensorflow as tf
```

```python
# 1) Fashion MNIST 로딩 + 전처리
#    (Input(28,28) 방식 — reshape 없이)
(train_input, train_target), (test_input, test_target) = \
    keras.datasets.fashion_mnist.load_data()
    
train_scaled = train_input / 255.0
```

```python
# 2) Flatten + Dense(100, relu) + Dense(10, softmax)
model = keras.Sequential([
    keras.layers.Input(shape=(28, 28)),
    keras.layers.Flatten(),
    keras.layers.Dense(100, activation='relu', name='hidden'),
    keras.layers.Dense(10, activation='softmax', name='output')
])

model.summary()
```

```python
# 3) compile(optimizer='adam',
#            loss='sparse_categorical_crossentropy')
model.compile(optimizer='adam', 
              loss='sparse_categorical_crossentropy', 
              metrics=['accuracy'])
```

```python
# 4) fit(epochs=5)
history = model.fit(train_scaled, train_target, 
                    epochs=5,
                    validation_split=0.2)
```

```python
# 5) evaluate → val_accuracy 확인
# val_scaled = train_scaled[int(len(train_scaled) * 0.8):]  # 전체의 80% 지점 인덱스
# val_target = train_target[int(len(train_target) * 0.8):]  # -> 여기서부터 뒤가 validation (20%)

# model.evaluate(val_scaled, val_target)

final_val_acc = history.history['val_accuracy'][-1]
print(f"Final val_accuracy: {final_val_acc:.4f}")
```
